```python
message = "Welcome to Python Programming Class"
```

```python
print("Removing extra spaces: ", message.strip())
```

```python
print("Lowercase: ",message.lower())
```

```python
print("Uppercase:", message.upper())
```

```python
print("Title case: ", message.title())
```

```python
print("Replace 'Python' with 'Advanced Python': ", message.replace("Python","Advanced Python") )
```

```python
print("Starts with welcome: ", message.startswith("Welcome"))
```

```python
print("Ends with class: ", message.endswith("Class"))
```

```python
print("Count of 'o': ", message.count('o'))
```

```python
print('Position of programming: ', message.find("Programming"))
```

```python
print("Words: ",message.split())
```
